```python
#Import libraries to begin building the map
import folium
import pandas as pd
```

```python
#Create Map: Initialize a folium map object, setting the initial location and zoom level.
map_center = [47.6606905, -117.4225863]  # Center the map around Riverfront Park because most locations are downtown
my_map = folium.Map(location=map_center, zoom_start=15)
```

```python
#Insert dataset
df = pd.read_csv('FinalBusinessList.csv', encoding='latin-1')
```

```python
#View coloumns to add to the markers for each business on the map
df.columns
```

```python
#Create markers for each business
for index, row in df.iterrows():
    if pd.notnull(row['Latitude']) and pd.notnull(row['Longitude']):
        # Create a formatted HTML string for the popup
        popup_html = f"<strong>{row['Business Name']}</strong><br>"
        if pd.notnull(row['Address']):
            popup_html += f"Address: {row['Address']}<br>"
        if pd.notnull(row['Phone']):
            popup_html += f"Phone: {row['Phone']}<br>"
        if pd.notnull(row['Category']):
            popup_html += f"Category: {row['Category']}<br>"
        if pd.notnull(row['Website']):
            popup_html += f'<a href="{row["Website"]}" target="_blank">Website</a>'

        # Add the marker
        folium.Marker(
            location=[row['Latitude'], row['Longitude']],
            popup=folium.Popup(popup_html, max_width=300),
        ).add_to(my_map)
```

```python
# Display the map
display(my_map)
```

```python
#Save map as html link to input into SiteCore
my_map.save('bipoc_business_map.html')
```
